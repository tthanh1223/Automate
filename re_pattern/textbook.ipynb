```python
import re
phoneNumRegex = re.compile(r'\d\d\d-\d\d\d-\d\d\d\d')
# compile() : Compile a regular expression pattern, returning a Pattern object.
print(phoneNumRegex)
mo = phoneNumRegex.search('123-456-7893')
# search() : take string and return None if the regex pattern is not found in the string.
print(mo.group())
# call group() on Match object 'mo' to return the match
```

1. Import the regex module with 'import re'
2. Create a 'Regex' object with re.compile() function. (raw string)
3. Pass the string to search into the `Regex` objects `search()` method. This returns a `Match` object.
4. Call the `Match` object’s `group()` method to return a string of the actual matched text.

```python
import re
phoneNumRegex = re.compile(r'(\d\d\d)-(\d\d\d-\d\d\d\d)')
mo = phoneNumRegex.search('123-456-7893')
print(mo.group(1))
print(mo.group(2))
print(mo.group(0))
print(mo.group())
print(mo.groups())
# groups() returns a tuple 
```

`()` have a special meaning in re, escape the ( and ) character with a backslash `\`.
In `re`, characters like `. ^ $ * + ?` have special meaning so if we want to detect these characters, we need to escape them with a backslash `\` like `\. \^ \$`

The `|` character is called a _pipe_. Can be used anywhere to match one of many expressions.
For example, `r'Batman|Tina Fey` will match either 'Batman' or 'Tiny Fey'. When both occur, the first occurrence of matching text will be returned as the `Match` object.

```python
import re
batRegex = re.compile(r'Bat(wo)?man')
mo1 = batRegex.search('The Batman begins')
print(mo1.group())

mo2 = batRegex.search('The Batwoman begins')
print(mo2.group())
```

The `(wo)?` part of re means that the pattern _wo_ is an optional group.

```python
import re
batRegex = re.compile(r'Bat(wo)*man')
mo1 = batRegex.search('The Batman begins')
print(mo1.group())

mo2 = batRegex.search('The Batwoman begins')
print(mo2.group())
mo3 = batRegex.search('The Batwowowowowoman begins')
print(mo3.group())
```

The `*` means "match zero or more" - the group that precedes the star can occur any number of times in the text.

```python
import re
batRegex = re.compile(r'Bat(wo)+man')
mo1 = batRegex.search('The Batman begins')
print(mo1)

mo2 = batRegex.search('The Batwoman begins')
print(mo2.group())
mo3 = batRegex.search('The Batwowowowowoman begins')
print(mo3.group())
```

The `+` means "match one or more" - unlike `*`, not require its group to appear in the matched string, the group preceding a plus must appear _at least once_. Or it will return `NoneType` like `mo1`.

Use `(Ha){3}` to match the string 'HaHaHa' but it won't match 'HaHa', or we can specify a range by writing a minimum, a comma and a maximum in between the braces `{}`. Like `(Ha){3,5}` will match 'HaHaHa', 'HaHaHaHa' and 'HaHaHaHaHa'.

Python re is _greedy_ by default, so they will match the longest string possible. The _non-greedy_ version of the braces, which matches the shorted string possible has the closing brace followed by a question mark `?`.

```python
import re
greedy = re.compile(r'(Ha){3,5}')
mo1 = greedy.search('HaHaHaHaHa')
print(mo1.group())
non_greedy = re.compile(r'(Ha){3,5}?')
mo2 = non_greedy.search('HaHaHaHaHa')
print(mo2.group())
```

`Regex` objects also have a `findall()` method, while `search()` will return a `Match` object of the first matched text in the search string, the `findall()` method will return the strings of every match in the search string.

```python
import re
phone = re.compile(r'\d\d\d-\d\d\d-\d\d\d\d')
mo1 = phone.search('123-456-7893')
print(mo1.group())

print(phone.findall('Cell: 415-555-9999 Work: 212-555-0000'))
phone2 = re.compile(r'(\d\d\d)-(\d\d\d)-(\d\d\d\d)')
print(phone2.findall('Cell: 415-555-9999 Work: 212-555-0000'))
```

If there are groups in the re, then `findall()` will return a list of tuples.

`\d` : Any numeric digit from 0 to 9.

`\D` : Any character that is not a numeric digit from 0 to 9.

`\w` : Any letter, numeric digit or the underscore character.

`\W` : Any character that is not a letter, numeric digit or the underscore `_` character.

`\s` : Any space, tab or newline character.

`\S` : not a space, tab or new line

```python
xmas = re.compile(r'\d+\s\w+')
a = xmas.findall('12 drummers, 11 pipers, 10 lords, 9 ladies, 8 maids, 7 swans, 6 geese,'
             '5 rings, 4 birds, 3 hens, 2 doves, 1 partridge')
print(a)
```

The re `\d+\s\w+` will match text that has one or more numeric digits (\d+) followed by a whitespace character, followed by one or more letter/digit/underscore.

We can define our own character class with `[]`. We can also include ranges of letters or number by using a hyphen `-`. And inside the square brackets, normal regular expression symbols are not interpreted as such. That means we don't need to escape the `., *, ?` with `\`.

By placing a caret character (`^`) just after the opening bracket, we can make a _negative character class_. Match all the characters that are _not_ in the character class.

```python
import re
consonant = re.compile(r'[^aeiouAEIOU]')
print(consonant.findall('Hello, world!'))
```

We can use `^` at the start of a regex to indicate that a match must occur at the _beginning_ of the searched text. Likewise, we can use `$` at the end of the regex to indicate the string must end with this regex pattern.

```python
import re
beginsWithHello = re.compile(r'^Hello')
print(beginsWithHello.search('Hello, world!'))
print(beginsWithHello.search('Hello, world!').group())
print(beginsWithHello.search('Hey, Max. Hello,everyone!'))
```

The `.` (or _dot_) character in a re is called a __wildcard__ and will match any character except for a newline.

```python
import re
atRegex = re.compile(r'.at')
lst = atRegex.findall('The cat in the hat sat on the flat mat.')
print(lst)
```

The `dot` character will match just one character, which is why the match for the text _flat_ in the previous example matched only `lat`. To match a dot, escape the dot with a `\`.

```python
nameRegex = re.compile(r'First Name: (.*) Last Name: (.*)')
mo = nameRegex.search('First Name: Vo Thai Last Name: Thanh')
print(mo.group(1))
print(mo.group(2))
print(mo.group())
```

We use the dot-star `(.*)` to stand in for that "anything." The dot-star uses _greedy_ mode: It will always try to match as much text as possible. To match any all text in a _non_greedy_ fashion, use the dot, star and question mark `(.*?)`. Like with braces, the question mark tells Python to match in a non-greedy way.

```python
nongreedyRegex = re.compile(r'<.*?>')
mo = nongreedyRegex.search('<html> <head> I love you </head> <body> I hate you </body> </html>')
mo3 = nongreedyRegex.findall('<html> <head> I love you </head> <body> I hate you </body> </html>')
print(mo.group())
print(mo3)
greedyRegex = re.compile(r'<.*>')
mo1 = greedyRegex.search('<html> <head> I love you </head> <body> I hate you </body> </html>')
mo2 = greedyRegex.findall('<html> <head> I love you </head> <body> I hate you </body> </html>')
print(mo1.group())
print(mo2)
```

The dot-star match everything except a newline. By passing `re.DOTALL` as the second argument to `re.compile()`, you can make the dot character match _all_ characters, including the newline character.

```python
noNewlineRegex = re.compile(r'.*')
mo = noNewlineRegex.search('Serve the public trust.\n'
                           'Protect the innocent.\n'
                           'Uphold the law.')
print(mo.group())

newlineRegex = re.compile(r'.*',re.DOTALL)
mo = newlineRegex.search('Serve the public trust.\nProtect the innocent.\nUphold the law.')
print(mo.group())
```

RE match text with the exact casing you specify. But can make the regex case-insensitive, we can pass `re.IGNORECASE` or `re.I` to `re.compile()`.

```python
robocop = re.compile(r'robocop', re.IGNORECASE)
mo = robocop.search('RoboCop is part man, part machine, all cop.').group()
print(mo)
mo1 = robocop.search('ROBOCOP protects the innocent.').group()
print(mo1)
mo2 = robocop.search('Al, why does your programming book talk about robocop so much?').group()
print(mo2)
```

```python
namesRegex = re.compile(r'Agent \w+')
mo = namesRegex.sub('CENSORED','Agent Alice gave the secret documents to Agent Bob.')
print(mo)
```

The `sub()` method for `Regex` objects is passed two arguments. The first argument is a string to replace any matches. The second is the string for the regular expression. The `sub()` method returns a string with the substitutions applied.

Sometimes you may need to use the matched text itself as part of the substitution. In the first argument to `sub()`, you can type `\1 \2, \3` and so on, to mean "Enter the text of group 1,2,3, and so on, in the substitution."

```python
agentNamesRegex = re.compile(r'Agent (\w)\w*')
string = agentNamesRegex.sub(r'\1****', 'Agent Alice told Agent Carol that Agent Eve knew Agent Bob was a double agent.')
print(string)
```

We can censor the name of the secret agents by showing just the first letters of their names. To do this, you could use the regex `Agent (\w)w*` and pass `r'\1****'` as the first argument to `sub()`. The `\1` in that string will be replaced by whatever text was matched by group `1---` that is.

With complex regexes, we tell the `re.compile()` function to ignore whitespace and comments inside the regular expression string. This "verbose mode" can be enabled by passing the variable `re.VERBOSE` as the second argument.

```python
# instead of typing like this - very fucking bad for your eyes
phoneRegex = re.compile(r'((\d{3}|\(\d{3}\))?(\s|-|\.)?\d{3}(\s|-|\.)\d{4}(\s*(ext|x|ext.)\s*\d{2,5})?)')
phoneRegex1 = re.compile(r"""(
        (\d{3}|\(\d{3}\))?           # area code
        (\s|-|\.)?                   # separator
        \d{3}                        # first 3 digits
        (\s|-|\.)                    # separator
        \d{4}                        # last 4 digits
        (\s*(ext|x|ext.)\s*\d{2,5})? # extension
        )""", re.VERBOSE)
```
